```python
import simex
from simex.engine import PaperTradingSimulatorCore
from simex.lob_event import NormalizedLobEvent
from simex.sink import InMemoryLogSink
from simex.types import NoAggressorNeededSentinel, Side, UpdateSource, UpdateType
```

```python
eng = PaperTradingSimulatorCore()
sink = InMemoryLogSink()
eng.set_log_sink(sink)
```

```python
ev = NormalizedLobEvent(
    1,
    2,
    Side.BUY,
    UpdateType.ADD,
    100,
    5,
    42,
    7,
    NoAggressorNeededSentinel,
    UpdateSource.HISTORICAL,
    "BTC-USDT",
)
```

```python
eng.update(ev)
```

```python
print("Depth @100:", eng.depth_at(Side.BUY, 100))
print("Top 1 bids:", eng.l2_top_n(Side.BUY, 1))

# See emitted fills (none here because it didn’t cross)
print("Fills:", sink.get_fills())
```

```python
cfg = simex.replay.ReplayConfig(require_monotonic_ts_received=True, fail_fast=False)
replay_session = simex.replay.ReplaySession(engine=eng)
```

```python
summary = replay_session.run([ev], config=cfg)
summary
```
